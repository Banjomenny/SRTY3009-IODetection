```python
# Code 1: Download CICIDS-2017 dataset
!pip install kagglehub -q

import kagglehub, glob, os, shutil

print("Downloading CICIDS-2017 from Kaggle...")
path = kagglehub.dataset_download("yashpotdarpy/cicids-2017")
print(f"Downloaded to: {path}")

# List the CSV files found
csv_files = sorted(glob.glob(os.path.join(path, '**/*.csv'), recursive=True))
print(f"\nFound {len(csv_files)} CSV files:")
for f in csv_files:
    size_mb = os.path.getsize(f) / 1024 / 1024
    print(f"  {os.path.basename(f):45s} {size_mb:.1f} MB")
```

```python
#Save a combined CSV to Google Drive (so you never download again):
```

```python
# Code 2: Save to Google Drive
from google.colab import drive
import pandas as pd
import numpy as np

# Mount your Google Drive
drive.mount('/content/drive')

SAVE_PATH = '/content/drive/MyDrive/AI_Security_A3/CICIDS2017_combined.csv'
os.makedirs(os.path.dirname(SAVE_PATH), exist_ok=True)

print("Loading and combining all CSV files...")
frames = [pd.read_csv(f, low_memory=False) for f in csv_files]
df = pd.concat(frames, ignore_index=True)
print(f"Combined shape: {df.shape[0]:,} rows × {df.shape[1]} columns")

print(f"\nSaving to Google Drive...")
df.to_csv(SAVE_PATH, index=False)
print(f"✅ Saved to: {SAVE_PATH}")
print(f"   File size: {os.path.getsize(SAVE_PATH)/1024/1024:.1f} MB")
```
